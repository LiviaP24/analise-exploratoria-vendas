### Análise Exploratoria de Vendas

```python
import pandas as pd
import numpy as np
from pathlib import Path

caminho = Path("../data/vendas_loja.csv")

df_vendas = pd.read_csv(caminho, encoding= 'utf-8', low_memory= False)

display(df_vendas.head(11))
```

#### Inspeção dos dados
1. Número de linhas | colunas


```python
display(df_vendas.shape)
```

2. Colunas com valores NULL: data_venda, produto, preco_unitario, regiao

```python
df_vendas.info()
```

3. Quantos registos estão duplicados?

```python
print('Duplicatas =',df_vendas.duplicated().sum())

display(df_vendas[df_vendas.duplicated(keep=False)].sort_values('id_linha'))
```

#### Tratamento dos dados

1. Fazer uma cópia dos dados

```python
df_temp = df_vendas

display(df_temp.head(11))
```

2. Remover NULL

```python
df_temp = df_temp.dropna()

display(df_temp)
```

3. Remover duplicatas

```python
df_temp = df_temp.drop_duplicates(keep='first')

display(df_temp)
```

Conversão de tipo

```python
df_temp['data_venda'] = pd.to_datetime(df_temp['data_venda'], format='%Y-%m-%d', errors='coerce')

datas_com_erro = df_temp.loc[df_temp['data_venda'].isna()]

display(df_temp[df_temp['data_venda'].isna()])
```
