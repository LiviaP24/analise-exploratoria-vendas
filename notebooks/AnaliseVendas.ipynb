### Análise Exploratoria de Vendas

```python
import pandas as pd
import numpy as np
from pathlib import Path

caminho = Path("../data/vendas_loja.csv")

df_vendas = pd.read_csv(caminho, encoding= 'utf-8', low_memory= False)

display(df_vendas.head(11))
```

#### Inspeção dos dados
1. Número de linhas | colunas


```python
display(df_vendas.shape)
```

2. Colunas com valores NULL: data_venda, produto, preco_unitario, regiao

```python
df_vendas.info()
```

3. Quantos registos estão duplicados?

```python
print('Duplicatas =',df_vendas.duplicated().sum())

display(df_vendas[df_vendas.duplicated(keep=False)].sort_values('id_linha').reset_index(drop = True).head(5))
```

#### Tratamento dos dados

1. Fazer uma cópia dos dados

```python
df_analise = df_vendas.copy()
```

2. Remover NULL

```python
df_analise = df_analise.dropna(subset=["data_venda", "produto", "quantidade", "preco_unitario", "status_pedido"])

display(df_analise.head(5))
```

3. Remover duplicatas

```python
df_analise = df_analise.drop_duplicates(keep='first')

display(df_analise.head(5))
```

Removendo valores negativos na coluna "quantidades"

```python
df_analise = (df_analise[df_analise["quantidade"] > 0].reset_index(drop=True))
```

Conversão de tipo

Foram encontradas datas incoerentes durante a conversão da coluna data_venda para o tipo datetime. Os valores inválidos foram transformados em NaT por meio do parâmetro errors="coerce". Em seguida, foi verificado se existiam outras linhas do mesmo pedido com uma data válida, permitindo recuperar essas informações sem criar valores arbitrários. Os registros originalmente sem data foram removidos anteriormente, pois não havia informação suficiente para recuperá-los com segurança.

```python
df_analise['data_venda'] = pd.to_datetime(df_analise['data_venda'], format='%Y-%m-%d', errors='coerce')

df_analise["data_venda"] = (df_analise["data_venda"].fillna(df_analise.groupby("id_pedido")["data_venda"].transform("first")))

display(df_analise.isna().sum())
```

Análise final

```python
df_analise.info()

print('Duplicatas = ',df_analise.duplicated().sum())
```

Os registros sem informação de região foram preservados nas análises gerais, pois continham dados válidos de venda. Entretanto, foram desconsiderados exclusivamente na análise regional, uma vez que não havia informação suficiente para atribuí-los a uma região com segurança. Não foi realizada imputação com base na cidade, pois essa correspondência não estava documentada como regra do conjunto de dados.

4. Qual foi o faturamento líquido dos pedidos concluídos?

```python
df_concluidos = df_analise[df_analise['status_pedido'] == 'Concluído']
```

```python
df_faturamento_liquido = df_concluidos[['quantidade', 'preco_unitario', 'desconto_percentual','produto','data_venda', 'regiao', 'categoria']]

df_faturamento_liquido['Preco_bruto'] = df_faturamento_liquido['quantidade'] * df_faturamento_liquido['preco_unitario']

df_faturamento_liquido['preco_liquido'] = df_faturamento_liquido['Preco_bruto'] * (1 - df_faturamento_liquido["desconto_percentual"] / 100)

preco_liquido = df_faturamento_liquido['preco_liquido'].sum()

print(f"Faturamento líquido: R$ {preco_liquido:,.2f}")
```

5. Qual foi o ticket médio por pedido? 

```python
quantidade_pedidos = df_concluidos['id_pedido'].nunique()

ticket_medio = preco_liquido / quantidade_pedidos

print(f"Ticket médio: R$ {ticket_medio:,.2f}")
```

6. Quais produtos geraram mais receita?

```python
df_produtos_mais_receita = df_faturamento_liquido[['produto','preco_liquido']].groupby('produto').sum()

df_produtos_mais_receita = df_produtos_mais_receita.sort_values(by= 'preco_liquido', ascending= False)

df_produtos_mais_receita = df_produtos_mais_receita.reset_index()

df_produtos_mais_receita = df_produtos_mais_receita.rename(columns={'preco_liquido':'receita_liquida'})

display(df_produtos_mais_receita.style.format({"receita_liquida": "R$ {:,.2f}"}))
```

7. Quais meses tiveram maior faturamento?

```python
df_faturamento_liquido['Mes'] = df_faturamento_liquido['data_venda'].dt.month

df_meses_faturamento = df_faturamento_liquido[['Mes','preco_liquido']].groupby('Mes').sum()

df_meses_faturamento = df_meses_faturamento.reset_index()

nomes_meses = {
    1: "Janeiro",
    2: "Fevereiro",
    3: "Março",
    4: "Abril",
    5: "Maio",
    6: "Junho",
    7: "Julho",
    8: "Agosto",
    9: "Setembro",
    10: "Outubro",
    11: "Novembro",
    12: "Dezembro"
}

df_meses_faturamento["Mes"] = df_meses_faturamento["Mes"].map(nomes_meses)

df_meses_faturamento = df_meses_faturamento.sort_values(by='preco_liquido', ascending= False)

df_meses_faturamento = df_meses_faturamento.reset_index(drop= True)

display(df_meses_faturamento.head(5).style.format({'preco_liquido': 'R$ {:,.2f}'}))
```

8. Quais regiões ficaram abaixo da média?

```python
df_regional = (df_faturamento_liquido.dropna(subset=["regiao"]).copy())

df_regioes_media = df_regional[["regiao", "preco_liquido"]].groupby("regiao").sum()
```

```python
media_faturamento = df_regioes_media['preco_liquido'].mean()

df_regioes_abaixo_media = df_regioes_media[df_regioes_media['preco_liquido'] < media_faturamento]

df_regioes_abaixo_media = df_regioes_abaixo_media.sort_values(by='preco_liquido', ascending= False)

df_regioes_abaixo_media = df_regioes_abaixo_media.reset_index() 

df_regioes_abaixo_media = df_regioes_abaixo_media.rename(columns={'preco_liquido':'faturamento total'})

display(df_regioes_abaixo_media.style.format({'faturamento total': 'R$ {:,.2f}'}))
```

9. Qual categoria apresentou maior crescimento de 2024 para 2025?

```python
df_faturamento_liquido['ano'] = df_faturamento_liquido['data_venda'].dt.year

df_crescimento_categoria = df_faturamento_liquido[['categoria', 'preco_liquido', 'ano']].groupby(['categoria', 'ano'], as_index = False).sum()

df_crescimento_categoria = df_crescimento_categoria.pivot(index= 'categoria', columns= 'ano', values= 'preco_liquido')

df_crescimento_categoria['Crescimento'] = df_crescimento_categoria[2025] - df_crescimento_categoria[2024]

df_crescimento_categoria['Crescimento percentual'] = (df_crescimento_categoria['Crescimento'] / df_crescimento_categoria[2024]) * 100

df_crescimento_categoria = df_crescimento_categoria.sort_values(by='Crescimento percentual', ascending= False)

df_crescimento_categoria.columns.name = None

df_crescimento_categoria = df_crescimento_categoria.rename(columns={2024.0: "Faturamento 2024", 2025.0: "Faturamento 2025"})

df_crescimento_categoria = df_crescimento_categoria.reset_index()

display(df_crescimento_categoria.style.format({"Faturamento 2024": "R$ {:,.2f}", "Faturamento 2025": "R$ {:,.2f}", "Crescimento": "R$ {:,.2f}", "Crescimento percentual": "{:.2f}%"}))
```
